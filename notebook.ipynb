```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import matplotlib.pyplot as plt
```

```python
nltk.download("punkt")
nltk.download("stopwords")
```

```python
reviews = pd.read_csv("reviews.csv")
reviews.head()
```

```python
reviews.info()
```

```python
negative_reviews_tmp = reviews[reviews['score'].isin([1, 2])]['content']
negative_reviews_tmp = negative_reviews_tmp.reset_index(drop=True)
negative_reviews_tmp.head()
```

```python
negative_reviews_tmp.info()
```

```python
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    tokens = word_tokenize(text)
    
    filtered_tokens = [
        token.lower()
        for token in tokens
        if token.isalpha() and token.lower() not in stop_words
    ]
    
    preprocessed_text = ' '.join(filtered_tokens)
    return preprocessed_text
```

```python
preprocessed_reviews = negative_reviews_tmp.to_frame()
```

```python
preprocessed_reviews['review'] = preprocessed_reviews['content'].apply(preprocess_text)

preprocessed_reviews.head()
```

```python
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(preprocessed_reviews['review'])

tfidf_matrix
```

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

tfidf_df.head()
```

```python
clust_kmeans = KMeans(n_clusters=5, random_state=42)

predicted_labels = clust_kmeans.fit_predict(tfidf_matrix)

categories = predicted_labels.tolist()
```

```python
pd.Series(categories).value_counts()
```

```python
terms = vectorizer.get_feature_names_out()

topic_terms_list = []

for cluster_num in range(clust_kmeans.n_clusters):
    
    cluster_indices = [
        idx for idx, label in enumerate(categories)
        if label == cluster_num
    ]
    
    cluster_tfidf_sum = tfidf_matrix[cluster_indices].sum(axis=0)
    cluster_tfidf_scores = np.array(cluster_tfidf_sum).ravel()
    
    top_10_indices = cluster_tfidf_scores.argsort()[-10:][::-1]
    top_10_terms = [terms[i] for i in top_10_indices]
    
    print(f"\nCluster {cluster_num} - Top 10 terms:")
    print(top_10_terms)
    
    top_term_index = top_10_indices[0]
    
    topic_terms_list.append({
        'category': cluster_num,
        'term': terms[top_term_index],
        'frequency': cluster_tfidf_scores[top_term_index]
    })
```

```python
topic_terms = pd.DataFrame(topic_terms_list)

topic_terms
```
